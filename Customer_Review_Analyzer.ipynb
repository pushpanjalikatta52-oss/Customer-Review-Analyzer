```python
# ==========================================
# 🛍️ CUSTOMER REVIEW ANALYZER
# Using NLP + Transformers + Gradio
# ==========================================

!pip -q install transformers torch gradio

import gradio as gr
from transformers import pipeline

# Load sentiment analysis model
sentiment_model = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)


# Review analysis function
def analyze_review(review):

    if not review.strip():
        return "Please enter a customer review."

    try:
        result = sentiment_model(review[:512])[0]

        label = result["label"]
        confidence = result["score"] * 100

        if label == "POSITIVE":
            emotion = "😊 Positive"
        else:
            emotion = "😞 Negative"

        return f"""
{emotion}

📊 Confidence: {confidence:.2f}%

📝 Review:
{review}

🏷️ Model Label: {label}
"""

    except Exception as e:
        return f"Error: {str(e)}"


# ==========================================
# Gradio Interface
# ==========================================

with gr.Blocks(title="Customer Review Analyzer") as app:

    gr.Markdown("""
    # 🛍️ Customer Review Analyzer

    ### Analyze customer reviews using Natural Language Processing (NLP)

    Enter a customer review below to identify whether the
    review is positive or negative.
    """)

    review_input = gr.Textbox(
        label="📝 Customer Review",
        placeholder="Example: The product quality is excellent and delivery was fast!",
        lines=6
    )

    analyze_button = gr.Button("🔍 Analyze Review")

    result_output = gr.Textbox(
        label="📊 Analysis Result",
        lines=10
    )

    analyze_button.click(
        fn=analyze_review,
        inputs=review_input,
        outputs=result_output
    )

    gr.Markdown("""
    **Technologies:** Python | NLP | Transformers | PyTorch | Gradio
    """)


# Launch application
app.launch(share=True)
```
